## Un problema de JO-EL

```python
#JOEL https://jo-el.es/problem/abecegrama
alfabeto = "abcdefghijklmnñopqrstuvwxyz"

n = int(input())

if n == 0:
    print("NO")

else:
    correcto = True

    for i in range(n):
        palabra = input()

        letra_esperada = alfabeto[i % 27]

        if palabra[0] != letra_esperada:
            correcto = False

    if correcto:
        print("SI")
    else:
        print("NO")
```
